```python
import random

# Pet names categorized by type, gender, and personality
pet_name_data = {
    "dog": {
        "male": {
            "playful": ["Buddy", "Ziggy", "Dash"],
            "calm": ["Charlie", "Max", "Cooper"],
            "brave": ["Rocky", "Hunter", "Ace"]
        },
        "female": {
            "playful": ["Luna", "Zoe", "Coco"],
            "calm": ["Daisy", "Bella", "Molly"],
            "brave": ["Nala", "Roxy", "Storm"]
        }
    },
    "cat": {
        "male": {
            "playful": ["Milo", "Simba", "Felix"],
            "calm": ["Shadow", "Leo", "Oliver"],
            "curious": ["Whiskers", "Oreo", "Mochi"]
        },
        "female": {
            "playful": ["Chloe", "Lily", "Ginger"],
            "calm": ["Cleo", "Nala", "Misty"],
            "curious": ["Mittens", "Nova", "Pixie"]
        }
    }
}

# 🐾 Get user input
pet_type = input("🐾 What type of pet? (dog/cat): ").lower()
gender = input("🐶 Is your pet male or female? ").lower()
personality = input("✨ What's their personality? (playful/calm/brave/curious): ").lower()

# 🧠 Generate name
if pet_type in pet_name_data and gender in pet_name_data[pet_type] and personality in pet_name_data[pet_type][gender]:
    name_suggestions = random.sample(pet_name_data[pet_type][gender][personality], 3)

    print(f"\n🎉 Here are 3 name suggestions for your {personality} {pet_type}:")
    for name in name_suggestions:
        print(f" - {name} ")
else:
    print("\n😿 Oops! I don't have names for that combo yet. Try again with dog/cat and one of the listed traits.")
```
